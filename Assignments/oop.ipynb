```python
class Book:
    def __init__(self, title, author, book_no): #contractor #each object to have its own set of initial attributes
        self.title = title
        self.author = author
        self.book_no = book_no
        self.is_borrowed = False

    def borrow(self):
        self.is_borrowed = True

    def return_book(self):
        self.is_borrowed = False

    def __str__(self):
        status = 'Borrowed' if self.is_borrowed else 'Available'
        return f'Book: {self.title} by {self.author} (ID: {self.book_no}) - Status: {status}'
```

```python
class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []

    def borrowed_book(self, book):
        if book not in self.borrowed_books: #conditional statement
            self.borrowed_books.append(book)
            book.borrow()

    def return_book(self, book):
        if book in self.borrowed_books:
            self.borrowed_books.remove(book)
            book.return_book()

    def __str__(self):
        book_list = '\n'.join([f' -{book.title}' for book in self.borrowed_books])
        return f'Member: {self.name}  (ID: {self.member_id})\nBorrower Books:\n{book_list or ' None'}'
```

```python
class Library:
    def __init__(self):
        self.books = []
        self.members = []

    def add_book(self, book):
        self.books.append(book)

    def register_member(self, member):
        self.members.append(member)

    def find_book(self, book_no):
        for book in self.books:
            if book.book_no == book_no:
                return book
        return None
    def find_member(self, member_id):
        for member in self.members:
            if member.member_id == member_id:
                return member
        return None

    def borrow_book(self, member_id, book_no):
        member = self.find_member(member_id)
        book = self.find_book(book_no)

        if not member:
            return 'Member not found'
        if not book:
            return 'Book not found'
        if book.is_borrowed:
            return 'Book is already borrowed'
        member.borrow_book(book)
        return f"Book  '{book.title}' borrowed by {member.name}"
    
    def return_book(self, member_id, book_no):
        member = self.find_member(member_id)
        book = self.find_book(book_no)

        if not member:
            return 'Member not found'
        if not book:
            return 'Book not found'
        if book not in member.borrowed_books:
            return 'This member did not borrow this book'
        member.return_book(book)
        return f"Book '{book.title}' returned by {member.name}"
    
    def __str__(self):
        books_info ='\n'.join([str(book) for book in self.books]) or 'No books in library'
        members_info ='\n'.join([str(member) for member in self.members]) or 'No members registered'
        return f"Library Status:\n\nBooks:\n{books_info}\n\nMembers:\n{members_info}"
```

```python
# Main program with user input
def main():
    library = Library()

    while True:
        print('\n==== Library Managment System ====')
        print('1. Add a new book')
        print('2. Register a new member')
        print('3. Borrow book')
        print('4. Return a book')
        print('5. Display library status')
        print('6. Exit')

        choice = input('Enter your choice (1-6) : ')

        if choice == '1':
            print('\n--- Add New Book ---')
            title = input('Enter book title: ')
            author = input('Enter author name: ')
            book_no = input('Enter book ID: ')
            # Check if book Id already exsits
            if library.find_book(book_no):
                print(f"Error: Book with ID '{book_no}' already exsits ")
            else:
                new_book = Book(title, author, book_no)
                library.add_book(new_book)
                print(f"Book '{title}' added successfully")

        elif choice == '2':
            print('\n--- Register New Member ---')
            name = input('Enter member name: ')
            member_id = input('Enter member ID')
            # Check if member already exsits
            if library.find_member(member_id):
                print(f"Error: Member with ID '{member_id}' already exists!")
            else:
                new_member = Member(name, member_id)
                library.register_member(new_member)
                print(f"Member '{name}' registered successfully")

        elif choice == '3':
            print('\n--- Borrow a Book ---')
            if not library.members:
                print('Error: No members registered in the library!')
                continue
            member_id = input('Enter your member ID: ')
            book_no = input('Enter book ID to borrow: ')
            result = library.borrow_book(member_id, book_no)
            print(result)

        elif choice == '4':
            print('\n--- Return a Book ---')
            if not library.members:
                print('Error: No members registered in the library! ')
                continue
            member_id = input('Enter your member ID: ')
            book_no = input('Enter book ID to Return: ')
            result = library.return_book(member_id, book_no)
            print(result)

        elif choice == '5':
            print('\n--- Current Library Status ---')
            print(library)

        elif choice == '6':
            print('Exiting Library Management System. Goodbye!')
            break

        else:
            print('Invalid choice, Please enter a number between 1-6.')

if __name__ == '__main__':
    main()

        
```
